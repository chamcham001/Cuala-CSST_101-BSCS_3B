# **3. Numeric Operation**

```r
x <- 5
y <- 3

sum <- x + y
dif <- x - y
mul <- x * y
div <- x / y

print (paste("Sum:", sum))
print (paste("dif:", dif))
print (paste("mul:", mul))
print (paste("div:", div))

cat("---------------------------","\n")

cat("+", sum, "\n")
cat("-", dif, "\n")
cat("*", mul, "\n")
cat("/", div, "\n")
```

# **8. Vectors**

```r
grades <- c(90,88, 96, 83,91)

grades

cat("-------------------------", "\n")

cat("Mean", mean(grades), "\n")
cat("Max", max(grades), "\n")
cat("Min", min(grades), "\n")
cat("Sum", sum(grades), "\n")
cat("Length", length(grades), "\n")
```
