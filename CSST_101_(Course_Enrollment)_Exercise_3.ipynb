# **1. Create the student database**

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005", "S006", "S007", "S008"),
  Programming_Grade = c(88, 72, 91, 78, 65, 85, 74, 82),
  Math_Grade = c(85, 70, 88, 80, 68, 90, 73, 79),
  Attendance = c(92, 80, 95, 88, 70, 91, 76, 85),
  Prerequisite = c("Yes", "No", "Yes", "Yes", "No", "Yes", "No", "Yes")
)

print(students)
```

# **2. Explore the data**

```r
summary(students)
nrow(students)
ncol(students)

cat("\n")

max(students$Programming_Grade)
max(students$Attendance)
max(students$Math_Grade)

cat("\n")

mean(students$Programming_Grade)
mean(students$Attendance)
mean(students$Math_Grade)

cat("\n")

min(students$Programming_Grade)
min(students$Attendance)
min(students$Math_Grade)
```

# **3. Enter a Student ID**

```r
student_id <- readline("Enter Student ID: ")
print(student_id)
```

# **4. Search the Data Frame**

```r
student <- students[
  students$Student_ID == student_id,
]
student
```

# **5. Check if the students Exists**

```r
if (nrow(student) > 0) {
  print("Found")
} else {
  print("Not Found")
}

student
```

# **6. Extract Student Information**

```r
programming <- student$Programming_Grade

cat(paste0("Student ID: ", student_id, "\n"))
cat(paste0("Programming Grade: ", programming))
```

# **7. Implement the Rules**

```r
student_eligibility <- function(programming, math, attendance, prerequisite)
if (programming >= 80 && math >= 75 && attendance >= 80 && prerequisite == "Yes"){
  print("Eligible")
}else if (programming >= 75 && math >= 70 && attendance >= 75){
  print("Conditionally Eligible")
}else{
 print("Not Eligible")
}

student_eligibility(78,87,54,"No")
```

# **8. Apply function to All Student**

```r
students$Eligibility <- mapply(
  student_eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite
)

students
```

# **9. Create Recommendation Rule**

```r
get_recommendation <- function(eligibility) {

  if (eligibility == "Eligible") {
    return("Student may Enroll")

  } else if (eligibility == "Conditional Eligible") {
    return("Consult Academic Adviser")

  } else {
    return("Complete requirements before enrollment")
  }
}

students$Recommendation <- sapply(
  students$Eligibility,
  get_recommendation
)

students
```

# **10. Expected Output**

```r
student_id <- readline(prompt = "Enter Student ID: ")

student <- students[students$Student_ID == student_id, ]

if (nrow(student) > 0) {

  cat("\n========================================\n")
  cat("       COURSE ENROLLMENT ADVISER\n")
  cat("========================================\n\n")

  cat("Student ID:", student$Student_ID, "\n\n")

  cat(sprintf("Programming Grade: %.2f%%\n", student$Programming_Grade))
  cat(sprintf("Math Grade: %.2f%%\n", student$Math_Grade))
  cat(sprintf("Attendance: %.2f%%\n", student$Attendance))
  cat("Prerequisite:", student$Prerequisite, "\n\n")

  cat("Enrollment Status:\n")
  cat(student$Eligibility, "\n\n")

  cat("Recommendation:\n")
  cat(student$Recommendation, "\n")

  cat("========================================\n")

} else {
  cat("\nStudent ID not found.\n")
}
```
