# **11. Creating a Simple Rules-Based System**

```r
classify_grade <- function(attendance, grade)
{
  if (attendance < 75 && grade < 75) {
    print("Student is at risk")
  } else {
    print("Student is not at risk")
  }
}

attendance <- 72
grade <- 70

classify_grade(attendance, grade)
```

# **12. Add More Rules**

```r
students_risk <- function (attendance, grade)

if (attendance < 75 && grade < 75) {
  print("High Risk")
} else if (attendance < 85 || grade < 80) {
  print("Moderate Risk")
} else if (attendance <= 85 && grade <= 80) {
  print("Low Risk")
} else {
  print("No Risk")
}
students_risk(96, 75)
```

# **13. Apply the Rule to Multiple Students**

```r
students$Risk <- mapply(
  students_risk,
  students$Attendance,
  students$Grade
)

students
```

# **16. Add a recommendation**

```r
recommendation <- function(risk) {

  if (risk == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk == "Moderate Risk") {
    return("Academic monitoring")
  } else if (risk == "Low Risk") {
    return("Continue regular monitoring")
  }
}

recommendation("High Risk")
```

# **17. Mini KRR System**

```r
student_id <- "S002"
attendance <- 72
grade <- 70

if (attendance < 75 && grade < 75) {
  risk <- "High Risk"
} else if (attendance < 85 || grade < 80) {
  risk <- "Moderate Risk"
} else {
  risk <- "Low Risk"
}

get_recommendation <- function(risk) {
  if (risk == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk == "Moderate Risk") {
    return("Academic monitoring")
  } else {
    return("Continue regular monitoring")
  }
}

recommendation <- get_recommendation(risk)

cat("Student ID:", student_id, "\n")
cat("Attendance:", attendance, "%\n")
cat("Grade:", grade, "\n\n")
cat("Risk Level:", risk, "\n\n")
cat("Recommendation:\n")
cat(recommendation, "\n")
```
